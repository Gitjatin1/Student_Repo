#### Write a function that determines if a student passed or failed based on their score.

```python
score=int(input("Enter student final score:-"))
def status (score):
    if score >= 60:
        print("This student is Pass")
    else:
        print("This stundent is Fail")
status(score)
```

#### Task 3: if-elif-else Statement
#### Write a function that converts a numeric grade to a letter grade:
#### 90-100: A, 80-89: B, 70-79: C, 60-69: D, Below 60: F

```python
score=int(input("Enter student final score:-"))
def status (score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >=60:
        return "D"
    else :
        return "F"
print("Student grade is:-",status(score))
```

#### Task 4: Nested Conditionals
#### Write a function to determine user access based on:
#### Login status (True or False)
#### Admin status (True or False)
#### Premium membership (True or False)
#### Conditions:
#### If not logged in, print "Please log in to access the site".
#### If admin, print "Welcome admin! Full access granted".
#### If premium user, print "Welcome premium user! Premium content access granted".
#### Otherwise, print "Welcome! Basic access granted".

```python
def access (log_stu,adm_stu,pre_memb):
    if log_stu:
        if adm_stu:
            print("Welcome admin! Full access granted.")
            if pre_memb:
                print("Welcome Premium user! Premium content access granted.")
        else:
            print("Welcome! Basic access granted.")
    else:
        print("Please log in to access the site.")
access(True, True, True)
print("")
access(True, True, False)
print("")
access(True, False, False)
print("")
access(False, False, False)
print("")
access(True, False, True)
print("")
access(False, True, False)
```

#### Task 6: Combining Conditions
#### Write a function to check if a year is a leap year.
#### A leap year is divisible by 4, except when divisible by 100 unless also divisible by 400.

```python
Year=int(input("Enter any year:-"))
def status (Year):
    if (Year % 4 == 0) and (Year % 100 != 0)or (Year % 400==0):
        print(Year,":-This is leap year")
    else:
        print(Year,":- This is not leap year")
status(Year)
```

#### Task 7: Best Practices (Refactoring)
#### Refactor the function below to be more readable and efficient:

```python
def classify_item(item_type,price,in_stock):
    result=""
    if item_type == "electronics":
        if price < 100:
            if in_stock:
                result = "Affodable electronics in stock."
            else:
                result = "Affodable electronics out of stock."
        else:
            if in_stock:
                result = "Premium electronics in stock."
            else:
                result = "Premium electronics out of stock."
    else:
        if price < 100:
            if in_stock:
                result = "Affodable non-electronics in stock."
            else:
                result = "Affodable non-electronics out of stock."
        else:
            if in_stock:
                result = "Premium non-electronics in stock."
            else:
                result = "Premium non-electronics out of stock."
    return result
classify_item("electronics",101,False)
```

```python
def inproved_class(item_type,price,in_stock):
    category="Electronics" if item_type == "Electronics" else "Non-Electronics"
    pricing="Affordable" if price < 100 else "Premium"
    stock_status="in stock" if in_stock else "out of stock"
    return f"{category},{pricing},{stock_status}"

inproved_class("Electronics",101,False)
```
