```python
cars={"Ford":['Falcon','Focus','Festiva','Fairlane'],
        "Holden":['Commodore','Captiva','Barina','Trailblazer'],
         "Nissan":['Maxima','Pulsar','350z','Navara'],
         "Honda":['Civic','Accord','Odyssey','Jazz'],
         "Jeep":['Grand Cherokee','Cherokee','Trailhawk','Trackhawk']}
```

```python
cars.values()
```

```python
for keys,values in cars.items():
    print(keys,values)
```

```python
car_list=list(cars.values())
print(car_list)
```

```python
focus=car_list[1][2]
print(focus)
```

```python
def get_all_jeep(cars=cars):
    """return a comma separated list of jeep models(original order)"""
    return ",".join(cars['Jeep'])
```

```python
print(get_all_jeep())
```

```python
def get_first_model_manu(cars=cars):
    """return a list of matching models (original ordering)"""
    return [models[2] for models in cars.values()]
```

```python
print(get_first_model_manu())
```

```python
def get_all_match_models(cars=cars, grep='trail'):
    results=[]
    for models in cars.values():
        for models in models:
            if grep.lower() in models.lower():
                results.append(models)
    return sorted(results)
```

```python
print(get_all_match_models())
```
