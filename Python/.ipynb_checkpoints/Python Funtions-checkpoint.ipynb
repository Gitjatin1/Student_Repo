#### Task 6: Function Composition
#### Create two functions:
#### celsius_to_fahrenheit: Converts Celsius to Fahrenheit using the formula:
#### F = C * 9/5 + 32
#### fahrenheit_to_celsius: Converts Fahrenheit to Celsius using the formula:
#### C = (F - 32) * 5/9

```python
def fahren (celsius):
    f=celsius * 9/5 + 32
    return f
def cels (fahrenheit):
    c=(fahrenheit-32) * 5/9
    return c

fahren(0), cels(32)
```

#### Task 7: Combining Functions and Conditionals
#### Write a function named analyze_temperature that:
#### Takes a temperature value and a unit ('C' or 'F').
#### Converts the temperature to the other unit.
#### Returns a description of the weather:
#### Below 0°C: "freezing"
#### 0-15°C: "cold"
#### 15-25°C: "moderate"
#### Above 25°C: "hot"

```python
def temp_stuts (t):
    if t <= 0:
        print("Temperature is below 0 degree it means its Freezing")
    elif t > 0 and t < 15:
        print("Temperature is between 0 to 15 degree it means its Cold")
    elif t > 15 and t < 25:
        print("Temperature is between 15 to 25 degree it means its Moderate")
    else:
        print("Temperature is above 25 degree it means its Hot")



def analyze_temperature(temp,unit):
    
    if unit.upper() == "F":
        print("Temperature is in Faherenheit:-", temp)
        #convet Faherenheit to Celsius
        cels=(temp-32) * 5/9
        print("convet into Celsius Temperature:-",cels)
        temp_stuts(cels)
    else:
        print("Temperature is in Celsius:-", temp)
        temp_stuts(temp)
 


analyze_temperature(32,"f")
```

#### Task 8: Function with Variable Arguments
#### Write a function named calculate_average that:
#### Accepts any number of numeric arguments.
#### Returns the average of all numbers provided.

```python
def calculate_average(*numbers):
    return sum(numbers)/len(numbers) 
calculate_average(1,2,3,4,5,6,7,8,9)
```

#### Task 9: Advanced Challenge - Student Grade Function
#### Write a function named calculate_final_grade that:
#### Takes a student's name and any number of assignment scores.
#### Calculates the average score.
#### Determines the letter grade:
#### 90+: A, 80-89: B, 70-79: C, 60-69: D, Below 60: F
#### Returns a formatted string:
#### Example: "Alice's final grade is 85 (B)."

```python
  
def calculate_final_grade(name,*score):
    if not score:
        print(name, "student has no score avaibale")
    else:
        avg_score=sum(score)/len(score)
    
    if avg_score >= 90:
        grade= "A"
    elif avg_score >= 80:
        grade= "B"
    elif avg_score >= 70:
        grade= "C"
    elif avg_score >=60:
        grade= "D"
    else:
        grade= "F"
    return f"{name}'s final grade is {avg_score:.1f} ({grade})."

print(calculate_final_grade("Jay",60,60,60,60,60))
```
