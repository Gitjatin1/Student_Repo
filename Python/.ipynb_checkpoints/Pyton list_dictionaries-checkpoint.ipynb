# Python Lists and Dictionaries - Practice Exercise

#### Create a shopping list with at least 5 items

```python
Shoping_list=["Apples","Bread","Eggs","Milk","Ice cream"]
print (f"My shoping list:{Shoping_list}")
```

#### Print out the third item in the list

```python
print(f"The third item on list is :{Shoping_list[2]}")
```

#### Replace the first item with a different item

```python
Shoping_list[0]="Cheese"
```

```python
print(f"Updating in list :{Shoping_list}")
```

#### Add an item to the end of the list

```python
Shoping_list.append("Banana")
```

```python
print(f"Add new items Banana in list :{Shoping_list}")
```

#### Create a dictionary of 3 grocery items with their prices

```python
Items={"Orange":2.99, "Cake":10.99, "Bagles":3.99}
print(f"list of grocery with pricing:{Items}")
```

#### Print a message showing one item and its price

```python
print(f"The price of cake is :{Items["Cake"]}")
```

#### Add a new item to the dictionary with its price

```python
Items["Bread"]=5.99
```

```python
print(f"Adding new item bread in list:{Items}")
```

#### Calculate and print the total cost of all items

```python
Total_sum=sum(Items.values())
```

```python
print(f"The total cost of all the items is:{Total_sum}")
```
